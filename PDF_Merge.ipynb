```python
import pandas as pd
import PyPDF2
```

```python
# Define the input and output paths
pdf1_path = 'input/1.pdf'
pdf2_path = 'input/2.pdf'
output_path = 'output/merged.pdf'
```

```python
def merge_pdfs(pdf1_path, pdf2_path, output_path):
    merger = PyPDF2.PdfMerger()
    
    # Append the first PDF
    merger.append(pdf1_path)
    
    # Append the second PDF
    merger.append(pdf2_path)
    
    # Write out the merged PDF
    merger.write(output_path)
    merger.close()
```

```python
# Call the function to merge
merge_pdfs(pdf1_path, pdf2_path, output_path)
```
